#### 1️⃣ List [ ]

##### **What is a List?**

A list is a built-in Python data structure used to store an ordered, mutable (changeable) collection of items. Lists can hold mixed data types (strings, integers, floats, etc.) and allow duplicate values.

##### **📝 Characteristics**
- Changed (Mutable)
- Orderd
- Allow dublicate
- Multi-type elements  
- Dynamic size

#### 2️⃣ Tuple ( )

##### **What is a Tuple?**

A tuple is a built-in Python data structure used to store an ordered collection of items. Unlike lists, tuples are immutable (meaning once created, their elements cannot be changed, added, or removed) and they are defined using parentheses ().

##### **📝 Characteristics**

- Unchangeable (Immutable)
- Ordered
- Allow dublication
- Multi-type elements

**Creating Tuple**

```python
names = ("mohamed", "hamza", "mariam")
print(names)
print(type(names))
```

```python
names = tuple(("mohamed", "hamza", "mariam"))
print(names)
print(type(names))
```

```python
numbers = tuple(range(0,10))
numbers
```

**Changing Elements**

A set is a built-in Python data structure used to store an unordered, mutable collection of unique elements. Sets do not allow duplicate values and are defined using curly braces {} or the set() constructor.

##### **📝 Characteristics**
- Changeable (Mutable)
- Unordered
- Not allow dublicate
- Multi-type elements
- Dynamic size

**Mathmatical and Logical Set Operators**

```python
# this operators done between sets only
```

```python
# .intersection() or "&" --> return new set containing the intersection elements between to sets
names1 = {"mohamed", "hamza", "mariam"}
names2 = {"mohamed", "salah", "selim"}
names = names1.intersection(names2)
names
```

```python
# .union()  or "|" --> return new set containing all elements of both, but remove duplication (because sets not allow duplicate)
names1 = {"mohamed", "hamza", "mariam"}
names2 = {"mohamed", "salah", "selim","selim"}
names = names1.union(names2) 
names
```

```python
# .difference() or "-" --> return new set containing the element found in the first not in the second set
names1 = {"mohamed", "hamza", "mariam"}
names2 = {"mohamed", "salah", "selim"}
names = names1.difference(names2)
names
```

```python
# .symmetric_difference() or "^" --> Returns a new set containing elements that are in either of the sets, but not in both (excludes the intersection).
names1 = {"mohamed", "hamza", "mariam"}
names2 = {"mohamed", "salah", "selim"}
names = names1.symmetric_difference(names2)
names
```

#### 4️⃣ Dictionary

##### **1. What is a Dictionary?**

A dictionary is a built-in Python data structure used to store data in key-value pairs. Dictionaries are ordered (as of Python 3.7+), mutable (changeable), and do not allow duplicate keys. They are defined using curly braces {} with colons : separating keys and values.

##### **📝 Characteristic**
- Changeable (Mutable)
- Ordered (In past was Unordered)
- Not allow dublicate key (must unique Key)
- Multi-data type
- Dynamic size

**📌 Important Note**

- The Key must be unique not dublicate
- if you dublicate the key it will override on the value that means the last value it will be saved with this key

**Creating Dictionary**

```python
student = {
    "name" : "mohamed",
    "age" : 20,
    "id" : 1234
}
student
```

```python
student = {
    "name" : "mohamed",
    "age" : 20,
    "id" : 1234,
}
print("The dict:",student)
print("The type:",type(student))
```

```python
# dictionary does not allow duplicate key so if you duplicate the it override on the value of this key
student = {
    "name" : "mohamed",
    "age" : 20,
    "id" : 1234,
    "name" : "hamza"
}
student
```

```python
# .fromkeys(iterable, ) --> it take iterable values to use them as key of new dictionary and you can add value assigned to all keys, by default none
key = ["name", "age", "id"]
my_dict = dict.fromkeys(key)
my_dict
```

```python
my_dict = dict.fromkeys(key, 30)
my_dict
```

```python
value = ("mohamed", 20, 1234)
my_dict = dict.fromkeys(key, value)
my_dict
```

**Changing**

```python
student = {
    "name" : "mohamed",
    "age" : 20,
    "id" : 1234,
}
student["name"] = "hamza"
student
```

**Accessing and Retrieving Elements**

```python
# .key() --> return all keys in the dictionary
student = {
    "name" : "mohamed",
    "age" : 20,
    "id" : 1234,
}
student.keys()
```

```python
# values() --> return all valuse in the dictionary
student = {
    "name" : "mohamed",
    "age" : 20,
    "id" : 1234,
}
student.values()
```

```python
# .items() --> return all keys and values pairs as list of tuples 
student = {
    "name" : "mohamed",
    "age" : 20,
    "id" : 1234,
}
student.items()
```

```python
# Using key return error if the key is not found
student = {
    "name" : "mohamed",
    "age" : 20,
    "id" : 1234,
}
student["age"]
```

```python
student = {
    "name" : "mohamed",
    "age" : 20,
    "id" : 1234,
} 
student["school"] # return error because key is not found
```

```python
# .get() --> return the value of specified key,  if the key is not found it return "none" instead raising error
student = {
    "name" : "mohamed",
    "age" : 20,
    "id" : 1234,
}
student.get("age")
```

```python
student = {
    "name" : "mohamed",
    "age" : 20,
    "id" : 1234,
}
student.get("school") # return none because the key is not found and get return that instead error
```

```python
# .setdefault() --> return value of specific key, if the key is not found it inserts this key with specified default value and the value by default is none
student = {
    "name" : "mohamed",
    "age" : 20,
    "id" : 1234,
}
print(student.setdefault("school"))
student
```

```python
student = {
    "name" : "mohamed",
    "age" : 20,
    "id" : 1234,
}
print(student.setdefault("club", "GWSC"))
student
```

**Adding Elements**

```python
# .update() --> Inserts specified items (key-value pairs) into the dictionary. If the key already exists, its value is updated (override on it)
student = {
    "name" : "mohamed",
    "age" : 20,
    "id" : 1234,
}
student.update({"club" : "GWSC", "university" : "mansoura"})
student
```

```python
student = {
    "name" : "mohamed",
    "age" : 20,
    "id" : 1234,
}
added_elements = {
    "club" : "GWSC",
    "university" : "mansoura"
}
student.update(added_elements)
student
```

**Removing**

```python
# .pop() --> pop the specified key and return it is value, if the key is not found it returns error
student = {
    "name" : "mohamed",
    "age" : 20,
    "id" : 1234,
}
popped_item = student.pop("id")
print(popped_item)
student
```

```python
student = {
    "name" : "mohamed",
    "age" : 20,
    "id" : 1234,
}
student.pop("club") # error the key is not found
student
```

```python
# .popitem() --> Removes and returns the last inserted key-value pair as a tuple. (In Python 3.7+, dictionaries maintain insertion order).
student = {
    "name" : "mohamed",
    "age" : 20,
    "id" : 1234,
}
popped = student.popitem()
print(popped)
student
```

```python
# .clear() --> remove all items of the dictionary
student = {
    "name" : "mohamed",
    "age" : 20,
    "id" : 1234,
}
student.clear()
student
```

```python
# del() --> remove the whole dictionary in default, if put key it delete this item, and if key is not found it returns error
student = {
    "name" : "mohamed",
    "age" : 20,
    "id" : 1234,
}
del(student["id"])
student
```
