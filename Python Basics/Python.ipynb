#### 1. What is a Variable?
A variable is a named reference or label that points to a value stored in your computer's RAM. Instead of interacting with complex raw memory addresses (like hexadecimal pointers), you use a human-readable name to easily reference, update, and manage data throughout your program.

#### 2. Python’s Object/Reference Model & Dynamic Typing
 - Python handles variables differently than compiled languages like C or Java:
 - Everything is an Object: Numbers, strings, functions, and modules are all objects residing in memory.
 - Variables are Pointers, Not Boxes: Writing x = 10 does not carve out a physical box named x containing 10. Instead, Python creates an integer        object representing 10 in memory and makes the variable x point to it.
 - Dynamic Typing: Variables themselves have no fixed data type; the objects they reference do. A variable can point to an integer on line 1 and a string on line 10 without throwing an error. 
 - if i reassigned the variable the pointer change to point the new value/object and the first one will be deleted by cach

```python
x ="mohamed"  # The x point to string object/value is mohamed
x = 10  # The poiter change to point integer object/value is 10
```

#### 4. Variable Names and Naming Rules
Python enforces strict syntax rules alongside community-accepted style guidelines (PEP 8):

- Mandatory Rules:

  - Must start with a letter or an underscore (_). It cannot start with a number (e.g., 1st_var is invalid).

  - Can only contain alpha-numeric characters and underscores.

  - Case-sensitive: loss and Loss are completely distinct variables.

  - Cannot be a Python reserved keyword (if, class, def, for, etc.).

- PEP 8 Style Guidelines:
 
 ![alt text](image-2.png)

  - Use snake_case for standard variables and functions (lowercase words separated by underscores, e.g., batch_size).

  - Use ALL_CAPS for constants (e.g., MAX_EPOCHS).

  - Avoid single-letter variable names except for temporary loop counters like i and j.

```python
# snake_case for standard variables and functions
user_total_score = 95
learning_rate = 0.001

#ALL_CAPS for constants (values that shouldn't change)
MAX_ITERATIONS = 100
API_TIMEOUT_SECONDS = 30

# 3. Case-Sensitivity
loss_value = 0.45
Loss_Value = 0.99  # This is a completely separate variable from loss_value!

print("Standard Variable (snake_case):", user_total_score)
print("Constant Value:", MAX_ITERATIONS)
print("Lowercase variable loss_value:", loss_value)
print("Uppercase variable Loss_Value:", Loss_Value)
```

#### 11. Common Mistakes to Avoid
- Using Keywords as Names: Naming a variable after a keyword (e.g., class = 5) raises an immediate syntax error.

- UnboundLocalError: Trying to modify a global variable inside a function without explicitly declaring it with the global keyword.

### 5️⃣ Operators


**1. Arithmetic Operators**

Used to perform mathematical calculations.

```python
# Arithmetic Operators Example
a = 10
b = 3

print(a + b)   # Addition: 13
print(a - b)   # Subtraction: 7
print(a * b)   # Multiplication: 30
print(a / b)   # Division: 3.3333333333333335
print(a // b)  # Floor Division (rounds down): 3
print(a % b)   # Modulus (remainder): 1
print(a ** b)  # Exponentiation (10 to the power of 3): 1000
```

**2. Assignment Operators**

Used to assign values to variables and update them.

```python
# Assignment Operators Example
x = 5       # Assigns 5 to x
print(x)

x += 3      # Equivalent to x = x + 3 (x becomes 8)
print(x)

x -= 2      # Equivalent to x = x - 2 (x becomes 6)
print(x)

x *= 4      # Equivalent to x = x * 4 (x becomes 24)
print(x)

x /= 3      # Equivalent to x = x / 3 (x becomes 8.0)
print(x)
```

**3. Comparison Operators**

Used to compare two values. They always return True or False.

```python
# Comparison Operators Example
a = 10
b = 20

print(a == b)   # Equal: False
print(a != b)   # Not equal: True
print(a > b)    # Greater than: False
print(a < b)    # Less than: True
print(a >= b)   # Greater than or equal to: False
print(a <= b)   # Less than or equal to: True
```

**4. Logical Operators**

Used to combine conditional statements.

```python
# Logical Operators Example
x = True
y = False

print(x and y)  # True only if BOTH are true: False
print(x or y)   # True if AT LEAST ONE is true: True
print(not x)    # Reverses the boolean state: False
```

**5. Identity Operators**

Used to compare whether two variables point to the exact same object in memory.

📌Important note
- "Is" check these point to same object
- "==" check these have same values


```python
# Identity Operators Example
a = [1, 2, 3]
b = [1, 2, 3]
c = a

print(a is c)     # True, because 'c' points to the exact same object as 'a'
print(a == b)
print(a is b)     # False, because they are separate objects in memory (even though contents match)
print(a is not b) # True
```

**6. Membership Operators**

Used to test if a sequence is presented in an object (like strings, lists, or tuples).

```python
# Membership Operators Example
my_list = [10, 20, 30, 40]
text = "Hello Python"

print(20 in my_list)     # True, 20 is inside the list
print(50 not in my_list) # True, 50 is not inside the list
print("Python" in text)  # True, substring exists in the text
```

**7. Bitwise Operators**

Used to compare (or perform low-level operations on) binary numbers (bits).


📌Impotrant Note

Bitwise operators work only for numbers data types

```python
# Bitwise Operators Example (using integers 10 and 4)
# 10 in binary = 1010
#  4 in binary = 0100

a = 10
b = 4

print(a & b)   # Bitwise AND: Sets each bit to 1 if both bits are 1 -> Output: 0 (0000)
print(a | b)   # Bitwise OR: Sets each bit to 1 if one of two bits is 1 -> Output: 14 (1110)
print(a ^ b)   # Bitwise XOR: Sets each bit to 1 if only one of two bits is 1 -> Output: 14 (1110)
print(~a)      # Bitwise NOT: Inverts all the bits -> Output: -11
print(a << 2)  # Zero fill left shift: Shifts bits left by pushing zeros in from the right -> Output: 40 (101000)
print(a >> 2)  # Signed right shift: Shifts bits right -> Output: 2 (10)
```

**8. Operator Precedence**
- Python evaluates operators based on specific priority rules (similar to standard mathematics, like multiplication before addition).

- Here is the general order from highest precedence (evaluated first) to lowest precedence (evaluated last):

```python
# Operator Precedence Reference & Example
# 1. Parentheses: ()
# 2. Exponentiation: **
# 3. Unary plus, minus, bitwise NOT: +x, -x, ~x
# 4. Multiplication, division, floor division, modulus: *, /, //, %
# 5. Addition and subtraction: +, -
# 6. Bitwise shifts: <<, >>
# 7. Bitwise AND: &
# 8. Bitwise XOR: ^
# 9. Bitwise OR: |
# 10. Comparisons, identity, membership: ==, !=, >, <, >=, <=, is, is not, in, not in
# 11. Logical NOT: not
# 12. Logical AND: and
# 13. Logical OR: or

# Example showing how parentheses change precedence:
result_without_parens = 10 + 3 * 2  # Multiplies first (3*2=6), then adds 10 -> Output: 16
result_with_parens = (10 + 3) * 2    # Adds first (10+3=13), then multiplies by 2 -> Output: 26

print(result_without_parens)
print(result_with_parens)
```

### 7️⃣ If Statement

**1. Introduction to the if Statement**

The if statement is used to make decisions in your code. It evaluates a condition (a boolean expression that results in True or False). If the condition is True, the indented block of code underneath runs. If it is False, the code is skipped.

```python
# Cell 1: Basic if statement example
age = 20

if age >= 18:
    print("You are an adult.")  # This will print because 20 >= 18 is True

print("This line always runs because it is outside the if block.")
```

**2. The if-else Statement**

What if you want something else to happen when the condition is not met? That is where else comes in. It acts as a fallback when the if condition evaluates to False.

📌Important Note
- The else statement must come last. You cannot have an elif after an else.
- You can also have an else without the elif.
- This makes it useful for error handling, validation, and providing default values.



```python
# Cell 2: if-else statement example
temperature = 15

if temperature > 25:
    print("It's a hot day outside.")
else:
    print("It's not hot today.")  # This will print because 15 > 25 is False
```

**3. The if-elif-else Statement (Multiple Conditions)**

When you have more than two possibilities, you can use elif (short for else if). Python checks conditions from top to bottom. As soon as it finds one that is True, it runs that block and skips the rest of the chain.

```python
# Cell 3: if-elif-else statement example
score = 85

if score >= 90:
    print("Grade: A")
elif score >= 80:
    print("Grade: B")  # This will execute because 85 >= 80 is True
elif score >= 70:
    print("Grade: C")
else:
    print("Grade: F")
```

**4. Nested if Statements**

You can place an if statement inside another if statement. This is called nesting, and it is used when you want to check a secondary condition only if the primary condition is first met.

```python
# Cell 4: Nested if statement example
is_logged_in = True
has_permission = True

if is_logged_in:
    print("Welcome back, user!")
    
    if has_permission:
        print("Access granted to the admin dashboard.")
    else:
        print("Access denied: Insufficient permissions.")
else:
    print("Please log in first.")
```

**5. Shorthand if (Inline Ternary Operator)**

If you only have a simple if-else statement that assigns a value or runs a single line, Python allows you to write it on a single line (often called a ternary conditional operator).

```python
# Cell 5: Shorthand if-else example
age = 20

status = "Adult" if age >= 18 else "Minor"

print(status)  # Output: Adult
```

📌Important Note
- Python can evaluate many types of values as True or False in an if statement.
  - Zero (0), empty strings (""), None, and empty collections are treated as False. 
  - Everything else is treated as True.This includes positive numbers (5), negative numbers (-3), and any non-empty string (even "False" is treated as True because it's a non-empty string).
